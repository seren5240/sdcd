```python
import torch
from sdcd.models.modules import SCCPowerIteration
```

```python
def dag_reg_power_grad(power_grad, adj_matrix):
    grad, A = power_grad.compute_gradient(adj_matrix)
    # with torch.no_grad():
    #     grad = grad - A * (grad * A).sum() / ((A**2).sum() + 1e-6) / 2
    # grad = grad + torch.eye(self.in_dim)
    h_val = (grad.detach() * A).sum()
    return h_val
```

```python
W_pred = torch.tensor([
    [ 0.,  1.3,  0., 0.],
    [ 0.,  0.,  -1.1, 0.],
    [ 3,  0.,  0., 0.],
    [ 0.,  0.,  0.,  0.],
], dtype=torch.float)

d = 4
power_grad = SCCPowerIteration(
                    torch.zeros(d, d, dtype=torch.float, requires_grad = True, device = 'cpu'),
                    d,
                )
sdcd_h = dag_reg_power_grad(power_grad, W_pred)
print(f'sdcd h: {sdcd_h}')
```
